# Neural Network Programs using TensorFlow Keras

This notebook demonstrates Multilayer Perceptron, Forward & Backward Propagation, and a simple Autoencoder using TensorFlow Keras.

## Multilayer Perceptron (MLP)

```python

import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
import numpy as np

# XOR dataset
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([[0], [1], [1], [0]])

# Build model
model = Sequential([
    Dense(4, input_dim=2, activation='sigmoid'),
    Dense(1, activation='sigmoid')
])

# Compile model
model.compile(optimizer='adam', loss='mse', metrics=['accuracy'])

# Train model
model.fit(X, y, epochs=5000, verbose=0)

# Evaluate and predict
predictions = model.predict(X)
print("Predictions:")
print(predictions.round())
```

## Forward and Backward Propagation (Simple)

```python

import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
import numpy as np

# Sample data
X = np.array([[0.5, 0.8]])
y = np.array([[1]])

# Build model
model = Sequential([
    Dense(1, input_dim=2, activation='sigmoid')
])

# Compile
model.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=0.1), loss='mse')

# Train model (backprop happens here)
model.fit(X, y, epochs=500, verbose=0)

# Predict (forward pass)
output = model.predict(X)
print(f'Predicted Output: {output[0][0]:.4f}')
```

## Simple Autoencoder

```python

import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
import numpy as np

# Identity input data
X = np.identity(3)

# Build Autoencoder model
autoencoder = Sequential([
    Dense(2, input_dim=3, activation='sigmoid'),  # Encoder
    Dense(3, activation='sigmoid')                # Decoder
])

# Compile model
autoencoder.compile(optimizer='adam', loss='mse')

# Train model
autoencoder.fit(X, X, epochs=5000, verbose=0)

# Predict reconstructed outputs
reconstructed = autoencoder.predict(X)
print("\nOriginal Input:")
print(X)
print("\nReconstructed Output:")
print(reconstructed.round(3))
```

```python
import numpy as np

X = np.array([[0.5, 1.2],
              [1.5, 2.3],
              [0.3, 0.7],
              [2.0, 3.5]])
y = np.array([[1.5],
              [3.0],
              [1.0],
              [4.5]])

np.random.seed(0)
W1 = np.random.randn(2, 3)
b1 = np.zeros((1, 3))
W2 = np.random.randn(3, 1)
b2 = np.zeros((1, 1))

def relu(x): return np.maximum(0, x)
def drelu(x): return (x > 0).astype(float)

for i in range(5000+1):
    Z1 = X @ W1 + b1
    A1 = relu(Z1)
    y_pred = A1 @ W2 + b2
    y_pred = relu(y_pred)
    
    loss = np.mean((y_pred - y)**2)
    
    d = 2 * (y_pred - y) / len(y)
    W2 -= 0.01 * (A1.T @ d)
    b2 -= 0.01 * np.sum(d, axis=0, keepdims=True)
    
    dz = (d @ W2.T) * drelu(Z1)
    W1 -= 0.01 * (X.T @ dz)
    b1 -= 0.01 * np.sum(dz, axis=0, keepdims=True)
    if i % 500 == 0:
        print(f"loss at epoch {i} is {loss}")


```

```python
import numpy as np
x = np.array([
    [1.5,0.2],
    [1.7,2.3],
    [3.2,1.3],
    [5.2,0.1]
])

y = np.array([
    [1.7],
    [1.3],
    [1.8],
    [2.3]
])

def relu(x):
    return np.maximum(0,x)

def relu_derivative(x):
    return (x > 0).astype(float)

lr = 0.001

# Weight initialization (scaled down)
first_layer_weight = np.random.randn(2,3) * 0.1
first_layer_bias = np.ones((1,3))

second_layer_weights = np.random.randn(3,1) * 0.1
second_layer_bias = np.ones((1,1))

for i in range(4000+1):
    first_layer_ip = (x @ first_layer_weight) + first_layer_bias
    first_layer_op = relu(first_layer_ip)

    second_layer_ip = (first_layer_op @ second_layer_weights) + second_layer_bias

    loss = np.mean((y - second_layer_ip)**2)

    dw2 = 2 * (y - second_layer_ip) / len(y)
    second_layer_weights -= lr * (first_layer_op.T @ dw2)
    second_layer_bias -= lr * np.sum(dw2, axis=0, keepdims=True)

    dw1 = (dw2 @ second_layer_weights.T) * relu_derivative(first_layer_ip)
    first_layer_weight -= lr * (x.T @ dw1)
    first_layer_bias -= lr * np.sum(dw1, axis=0, keepdims=True)

    if i % 500 == 0:
        print(f"loss at epoch {i} is {loss}")

```

```python
import numpy as np

# Data (now as numpy array and normalized)
x = np.array([
    [1.5,0.2],
    [1.7,2.3],
    [3.2,1.3],
    [5.2,0.1]
])
x = x / 5.2

y = np.array([
    [1.7],
    [1.3],
    [1.8],
    [2.3]
])

# Activation: tanh
def tanh(x):
    return np.tanh(x)

def tanh_derivative(x):
    return 1 - np.tanh(x)**2

# Initialize weights (small values)
first_layer_weight = np.random.randn(2,3) * 0.1
first_layer_bias = np.zeros((1,3))

second_layer_weights = np.random.randn(3,1) * 0.1
second_layer_bias = np.zeros((1,1))

lr = 0.01

for i in range(4000+1):
    # Forward
    first_layer_ip = (x @ first_layer_weight) + first_layer_bias
    first_layer_op = tanh(first_layer_ip)

    second_layer_ip = (first_layer_op @ second_layer_weights) + second_layer_bias

    loss = np.mean((y - second_layer_ip)**2)

    # Backward
    dw2 = 2 * (y - second_layer_ip) / len(y)
    second_layer_weights -= lr * (first_layer_op.T @ dw2)
    second_layer_bias -= lr * np.sum(dw2, axis=0, keepdims=True)

    dw1 = (dw2 @ second_layer_weights.T) * tanh_derivative(first_layer_ip)
    first_layer_weight -= lr * (x.T @ dw1)
    first_layer_bias -= lr * np.sum(dw1, axis=0, keepdims=True)

    if i % 500 == 0:
        print(f"loss at epoch {i} is {loss}")
```
